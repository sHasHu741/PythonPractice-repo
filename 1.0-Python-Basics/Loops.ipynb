## While Loops

The while loops continue to execute as long as the condition is true

```python
count=0

while count<5:
    print(count)
    count=count+1
```

```python
count=0

while count%2==0:
    print(count)
    count=count+1
```

## Loop Control Statements

### Break
The break statement exists the loop prematurely

```python
## break
for i in range(10):
    if i==5:
        break
    print(i)
```

### Continue

The continue statement skips the current iteration and continues with the next

```python
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

### Pass

The pass statement is a null operation; it does nothing

```python
for i in range(5):
    if i==3:
        print("The number is",i)
        pass
    print(i)
```

## Nested Loops
 
A loop inside a loop 

```python
## Nested loops

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}") ##USed f string to display in the above format in output
```

### Examples

Q. Calculate the sum of first n natural number using a while and for loop

```python
## While loop

n=int(input("Enter the value of n:"))
sum=0
count=1

while count<=n:
    sum=sum+count
    count=count+1
print("Sum of first",n,"natural numbers is:",sum)    

```

```python
## For loop

n=int(input("Enter the value of n:"))
sum=0

for i in range(n+1):
    sum=sum+i
print(sum)    

```

Q. Prime numbers between 1 and 100

```python
## For Loops

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)    
```
