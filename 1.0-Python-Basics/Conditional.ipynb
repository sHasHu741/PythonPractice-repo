# Conditional Statements

## If statement

if statement evaluates the condition and if the condition is true it prints the  blocks the code within it and gives the result.

```python
## If statement
age=20
if age>=18:
    print("You are allowed to vote in the elections")
```

## Else

The else statement executes the block of code if the condition in the if statement is false

```python
## Else statement
age=16
if age>=18:
    print("You are eligible to vote")
else:
    print("You are a minor")    
```

## Elif

The elif statement allow you to check multiple conditions. It stands for "else if"

```python
## Elif
age=20
if age<13:
    print("You are a child")
elif age<18:
    print("You are a teenager")
else:
    print("You are an adult") 
```

## Nested Conditional Statements

You can place one or more if, elif or else statement inside another if, elif or else statement to create an nested condition statement. 

Q Number whether is even, odd, negative

```python
num=int(input("Enter the number"))
if num>0:
    print("The number is positive")
    if num%2==0:
         print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is zero or negative")        
```

Q: Determine if a year is a leap year using nested condition statement

```python
year=int(input("Enter the year"))
if (year%4==0):
    if (year%100==0):
        if (year%400==0):
            print(year,"Year is a leap year")
        else:
            print(year,"Year is not a leap year")
    else:
        print(year,"Year is a leap year")
else:
    print(year,"Year is not a leap year")                       
```
