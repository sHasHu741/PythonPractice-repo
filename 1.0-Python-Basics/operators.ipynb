## Comparision Operators

These operators are used for comparision between two or more operators and result in either true or false

```python
## (==)Equal to Operator
a=10
b=10
print(a==b)
```

```python
str1="Shashwat"
str2="shashwat"
print(str1==str2)
```

```python
## (!=)Not equal to operator
print(str1!=str2)
```

```python
## Greater than(>)
num1=45
num2=55
print(num1>num2)
```

```python
## Less than(<)
num1=45
num2=55
print(num1<num2)
```

```python
## Greater than or equal to, Less than or equal to
num1=45
num2=55
print(num1>=num2)
print(num1<=num2) 
```

## Logical Operators

And: when both the operators are bollean true then only we get true otherwise we will get false
Or: will return true even if one is bollean true among the two only returns false when both statements are false
Not: it is inverse of any bollean variable i.e if any bollean variable is assigned true adding not makes the final result false

```python
## And, Not, OR
##And
X=True
Y=False
result=X and Y
print(result)

X=False
Y=True
result=X and Y
print(result)

X=False
Y=False
result=X and Y
print(result)

X=True
Y=True
result=X and Y
print(result)
```

```python
## OR

X=True
Y=False
result=X or Y
print(result)

X=False
Y=True
result=X or Y
print(result)

X=False
Y=False
result=X or Y
print(result)

X=True
Y=True
result=X or Y
print(result)
```

```python
## NOT
X=True
print(not X)

Y=False
print(not Y)
```
