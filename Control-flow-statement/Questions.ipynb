Q8. Write a program that asks the user to input the number until they input 0. The program should print the sum of all input numbers.

```python
total=0
while True:
    number=float(input("Enter the number(0 to stop):"))
    if number==0:
        break
    total=total+number
print("The sum of all the numbers is",total)
```

Q12. Write a program that calculates the factorial of a number input by the user using a while loop. 

```python
number=int(input("Enter the number:"))
factorial=1
i=1
while i<=number:
    factorial=factorial*i
    i=i+1
print("The factorial of the number",number,"is",factorial)    
```

Q13. Write a program that calculates the sum of the digits of a number input by the user using a while loop.

```python
number=int(input("Ener the number:"))
sum_of_digits=0
while number>0:
    digit=number%10
    sum_of_digits=sum_of_digits+digit
    number=number//10
print("The sum of digits of the is:",sum_of_digits)    
```

Q14. Write a program that checks if a number input by the user is a prime number using a for loop.

```python
num=int(input("Enter the number:")) 
is_prime=True
if num<=1:
    is_prime=False
else:
    for i in range(2,int(num**0.5)+1):
        if num%i==0:
            is_prime=False
            break
if is_prime:
    print(num,"is a prime number")
else:
    print(num,"is not a prime number")                
```

Q15. Write a program that prints the first n Fibonaccian numbers, where n is the input by the user.

```python
n=int(input("Enter the number of Fibonacci to be print by the user:"))
a,b=0,1
count=0
while count<n:
    print(a)
    a,b=b,a+b
    count=count+1
```
